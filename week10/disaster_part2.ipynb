```python
import pandas as pd
import matplotlib.pyplot as plt

# データ読み込み
df = pd.read_csv("emdat.csv")

# 確認
df.head()
```

```python
def count_disasters_by_year(data, year):
    """指定した年の災害件数を返す関数"""
    return len(data[data["Start Year"] == year])
```

```python
sample_year = 2000
print(f"{sample_year} 年の災害件数：", count_disasters_by_year(df, sample_year))
```

```python
years = range(1990, 2021, 5)
counts = [len(df[df["Start Year"] == y]) for y in years]

plt.figure(figsize=(7, 4))
plt.bar([str(y) for y in years], counts)
plt.title("5年ごとの災害件数")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

# 結果まとめ

関数により年ごとの災害件数を確認でき、Loop を用いて複数年の災害件数グラフを自動生成できた。
本分析により、災害件数の年ごとの変化を視覚的に理解することができる。
